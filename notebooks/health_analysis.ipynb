# Healthcare & Insurance Claims Exploratory Data Analysis
This notebook performs exploratory data analysis (EDA) on healthcare insurance claim data to identify key cost drivers using **pandas**, **matplotlib**, and **seaborn**.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid")

# Load insurance dataset
df = pd.read_csv('insurance.csv')
df.head()
```

```python
# Dataset inspection
print("Shape:", df.shape)
print("\nMissing Values:\n", df.isnull().sum())
print("\nData Types:\n", df.dtypes)
df.describe()
```

```python
# Statistical Breakdown by Smoker Status
smoker_stats = df.groupby('smoker')['charges'].agg(['count', 'mean', 'median', 'std']).reset_index()
smoker_stats
```

### Key Findings
1. **Smoker Status:** Smoker status is the primary multiplier for medical claim charges (~4x average charge increase).
2. **BMI Risk:** Obese individuals (BMI > 30) who smoke experience exponential charge spikes compared to non-smokers in similar BMI brackets.
